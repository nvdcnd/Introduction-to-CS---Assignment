```python
INDICATORS = {
    'EG.ELC.ACCS.ZS':    'Access to electricity (% of population)',
    'EG.CFT.ACCS.ZS':    'Access to clean fuels for cooking (% of population)',
    'IT.NET.USER.ZS':    'Individuals using the Internet (% of population)',
    'SP.URB.TOTL.IN.ZS': 'Urban population (% of total population)',
    'NY.GDP.PCAP.CD':    'GDP per capita (current US$)',
}

selected_indicators = list(INDICATORS.keys())

metadata = df[df['Series Code'].isin(selected_indicators)][
    ['Series Code', 'Topic', 'Indicator Name', 'Source', 'Periodicity', 'Aggregation method']
]

display(metadata)
```

## Data Cleaning
In this section should demonstrate how you have programmatically cleaned and prepared the dataset you have chosen.
Explain the steps you take.

```python
# Cell 6
print(df.shape)
print(df.columns.tolist())
print(df.dtypes)
display(df.head(3))
```

```python
#Cell 8
# 1. Xoá cột thừa ở cuối (Unnamed: 35)
# Dùng errors='ignore' để nếu lỡ chạy lại cell 2 lần cũng không bị báo lỗi
df_data = df.drop(columns=['Unnamed: 35'], errors='ignore')

# 2. Xác định danh sách 31 cột năm (từ cột thứ 5 trở đi, tức index 4 đến hết)
# 4 cột đầu là: Country Name, Country Code, Indicator Name, Indicator Code
year_cols = df_data.columns[4:] 

# 3. Chỉ ép kiểu numeric cho 31 cột năm (tránh ép nhầm 4 cột chữ đầu tiên)
for col in year_cols:
    df_data[col] = pd.to_numeric(df_data[col], errors='coerce')

# 4. Kiểm tra lại shape và kiểu dữ liệu sau xử lý
print("Shape sau khi xử lý:", df_data.shape) # Kỳ vọng: (106488, 35)
print("\nKiểu dữ liệu của vài cột năm sau khi ép:")
print(df_data[year_cols[:5]].dtypes)
```

```python
# cell 9
# 1. Đọc file SDGCountry.csv
df_country = pd.read_csv('Dataset\SDGCountry.csv')

# 2. Tạo cột cờ (flag): True nếu Region KHÔNG rỗng (quốc gia thật), False nếu rỗng (vùng tổng hợp)
df_country['is_country'] = df_country['Region'].notna()

print(df_country.columns.tolist())

# 3. Đếm số lượng từng nhóm
counts = df_country['is_country'].value_counts()
print(f"Số lượng quốc gia thật: {counts[True]}")      # Kỳ vọng: 217
print(f"Số lượng vùng tổng hợp: {counts[False]}")    # Kỳ vọng: 44

# 4. Lọc và in ra danh sách tên 44 vùng tổng hợp
aggregates = df_country[~df_country['is_country']]['Short Name'].tolist() 
# Lưu ý: Nếu file của bạn dùng cột 'Table Name' thay vì 'Short Name' thì đổi tên cột tương ứng nhé

print("\nDanh sách 44 vùng tổng hợp:")
for i, name in enumerate(aggregates, 1):
    print(f"{i}. {name}")
```

```python
# cell 11
# Luu y: 'Indicator Code' nam trong SDGData.csv, khong nam trong SDGCountry.csv
df_data = pd.read_csv('Dataset/SDGData.csv')

valid_country_codes = df_country.loc[df_country['is_country'], 'Country Code']

df_subset = df_data[
    df_data['Country Code'].isin(valid_country_codes)
    & df_data['Indicator Code'].isin(selected_indicators)
].copy()

print('Shape cua df_subset:', df_subset.shape)  # Ky vong: (217*5, 35) = (1085, 35)
display(df_subset.head(3))
```

```python
# -- CELL 12: khao sat missing theo NAM va theo CHI SO (subset 5 chi so) --
# df_subset dang o dang wide (cot nam 1990..2020), can doi sang long de co cot Year/Value
year_cols = [c for c in df_subset.columns if str(c).isdigit()]
id_cols = [c for c in ['Country Name', 'Country Code', 'Indicator Name', 'Indicator Code'] if c in df_subset.columns]

df_subset_long = df_subset.melt(
    id_vars=id_cols,
    value_vars=year_cols,
    var_name='Year',
    value_name='Value'
).copy()

df_subset_long['Year'] = pd.to_numeric(df_subset_long['Year'], errors='coerce')
df_subset_long['Value'] = pd.to_numeric(df_subset_long['Value'], errors='coerce')

coverage_year = df_subset_long.groupby('Year')['Value'].agg(non_null='count', total='size')
coverage_year['pct_thieu'] = (100 * (1 - coverage_year['non_null'] / coverage_year['total'])).round(1)
display(coverage_year)

coverage_ind = df_subset_long.groupby('Indicator Code')['Value'].agg(non_null='count', total='size')
coverage_ind['pct_thieu'] = (100 * (1 - coverage_ind['non_null'] / coverage_ind['total'])).round(1)
display(coverage_ind.sort_values('pct_thieu', ascending=False))

# [TODO] Ve hinh cho Quan sat 1: so gia tri CO DU LIEU theo nam
# Goi y: 2 duong — toan bo SDGData vs subset 5 chi so (do phu tang dan roi roi o 2020)
# Nguon chu thich: so lieu 2020 chiu anh huong COVID (trich SDGSeries-Time.csv)
```
